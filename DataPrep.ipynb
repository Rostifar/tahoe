```python
class TinyStoriesMeta:
    end_text_token = '<|endoftext|>'
    tiny_stories_train_path = 'data/TinyStoriesV2-GPT4-train.txt'
    tiny_stories_val_path = 'data/TinyStoriesV2-GPT4-valid.txt'
```

```python
def process_tiny_stories(path: str, max_examples: int | None = None):
    example_count = 0
    with open(path, 'r') as f:
        example = []
        for lineno, line in enumerate(f):
            line = line.strip()
            # leading whitespace
            if line == "" and not example:
                continue
            
            # assumption: EOT happens on newline
            if line == TinyStoriesMeta.end_text_token:
                if example:
                    yield "".join(example + [line])
                example_count += 1
                if example_count >= max_examples:
                    return
                example = []
            else:
                if TinyStoriesMeta.end_text_token in line:
                    print("Warning, unexpected EOT token found at line {}.", lineno)
                # newline separate following sentences                
                example.append("\n" + line if example else line)
    if example:
        print(f"Warning, unexpected hanging example found at end of text.")
```
