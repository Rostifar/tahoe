# Tokenization

```python
import os
import regex as re
```

```python
GPT2_PAT = r"""'(?:[sdmt]|ll|ve|re)| ?\p{L}+| ?\p{N}+| ?[^\s\p{L}\p{N}]+|\s+(?!\S)|\s+"""
re.findall(GPT2_PAT, "hello, world is the324")
```

```python
with open("../data/TinyStoriesV2-GPT4-train.txt", "rb") as f:
    f.seek(0, os.SEEK_END)
    file_size = f.tell()
    f.seek(0)

    data = f.read(file_size // 2)
    text = data.decode("utf-8")
    text = re.split("<|end_of_text|>", text)
    text = "".join(text)
    text = re.findall(GPT2_PAT, text)
```
